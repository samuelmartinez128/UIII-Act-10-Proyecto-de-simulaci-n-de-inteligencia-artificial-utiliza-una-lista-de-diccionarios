```python
# Inicialización de las estructuras de datos
minero = []
losmineros = {}

print("--- SISTEMA DE GESTIÓN MINERA IA (VERSIÓN TABLA) ---")

while True:
    print("\n" + "="*50)
    print("1. Registrar Minero")
    print("2. Mostrar Tabla de Datos")
    print("3. IA: Actualizar Niveles")
    print("4. Salir")

    opcion = input("\nSeleccione una opción: ")

    if opcion == "1":
        # Creamos el diccionario vacío para el nuevo registro
        minero = {}
        nombre = input("Nombre: ")
        mineral = input("Mineral (oro/plata/piedra): ")

        # Guardamos datos en el diccionario
        minero["nombre"] = nombre
        minero["mineral"] = mineral
        minero["nivel"] = 1

        # Agregamos el diccionario a la lista
        losmineros.append(minero)
        print(f"> {nombre} ha sido agregado a la base de datos.")

    elif opcion == "2":
        print("\n--- TABLA DE DATOS EN MEMORIA ---")
        if losmineros == []:
            print("LA LISTA ESTÁ VACÍA")
        else:
            # Encabezado de la tabla
            print(f"{'NOMBRE':<15} | {'MINERAL':<10} | {'NIVEL':<5}")
            print("-" * 35)

            # El ciclo FOR recorre cada diccionario en la lista
            for m in losmineros:
                # Accedemos a las llaves del diccionario para imprimir la fila
                print(f"{m['nombre']:<15} | {m['mineral']:<10} | {m['nivel']:<5}")

    elif opcion == "3":
        print("\n> IA Analizando rendimiento...")
        if losmineros == []:
            print("No hay datos que procesar.")
        else:
            for m in losmineros:
                if m["mineral"] == "oro":
                    m["nivel"] = m["nivel"] + 5
                elif m["mineral"] == "plata":
                    m["nivel"] = m["nivel"] + 2
                else:
                    m["nivel"] = m["nivel"] + 1
            print("¡Niveles actualizados según el valor del mineral!")

    elif opcion == "4":
        print("Cerrando sistema...")
        break

    else:
        print("Opción no válida.")
```
